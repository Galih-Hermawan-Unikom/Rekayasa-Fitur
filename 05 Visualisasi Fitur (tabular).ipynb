# Histogram

**Data Kehadiran Mahasiswa**

Langkah selanjutnya adalah membuat histogram untuk melihat distribusi frekuensi kehadiran mahasiswa. Histogram akan membantu kita memahami sebaran kehadiran dan apakah ada pola atau tren tertentu.

```python
# Membuat histogram kehadiran (tanpa filter)
plt.figure(figsize=(8, 6))
plt.hist(data['kehadiran'], bins=10, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Histogram Kehadiran Mahasiswa')
plt.xlabel('Kehadiran')
plt.ylabel('Frekuensi')
plt.grid(True)
plt.show()
```

```python
# Filter data untuk mata kuliah "Struktur Data"
data_struktur_data = data[data['mata_kuliah'] == 'Struktur Data']

# Buat histogram kehadiran mahasiswa
plt.figure(figsize=(10, 6))
plt.hist(data_struktur_data['kehadiran'], bins=20, color='skyblue', edgecolor='black')
plt.title('Histogram Kehadiran Mahasiswa di Mata Kuliah Struktur Data')
plt.xlabel('Kehadiran')
plt.ylabel('Jumlah Mahasiswa')
plt.grid(True)
plt.show()
```

**Histogram Jumlah Mahasiswa untuk Mata Kuliah yang Ditentukan**

```python
# Memilih mata kuliah tertentu
mata_kuliah_terpilih = 'Struktur Data'

# Filter data untuk mata kuliah tertentu
data_mata_kuliah_terpilih = data[data['mata_kuliah'] == mata_kuliah_terpilih]

# Membuat histogram distribusi nilai akhir
plt.figure(figsize=(8, 6))
plt.hist(data_mata_kuliah_terpilih['nilai_akhir'], bins=10, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Distribusi Nilai Akhir - {}'.format(mata_kuliah_terpilih))
plt.xlabel('Nilai Akhir')
plt.ylabel('Frekuensi')
plt.grid(True)
plt.show()
```

# Box Plot Nilai

Menggunakan box plot untuk melihat distribusi nilai tugas, UTS, dan UAS, serta untuk mengidentifikasi adanya pencilan (outliers).

Cara baca.



```python
# Membuat box plot untuk nilai tugas, UTS, dan UAS
plt.figure(figsize=(10, 6))
data[['nilai_tugas', 'nilai_uts', 'nilai_uas', 'nilai_akhir']].boxplot()
plt.title('Box Plot Nilai Tugas, UTS, UAS, dan Nilai Akhir Mahasiswa')
plt.ylabel('Nilai')
plt.xlabel('Mata Kuliah')
plt.show()
```

```python
# Filter data untuk mahasiswa yang mengambil mata kuliah "Pemrograman Berorientasi Objek"
mk1 = 'Pengantar Informatika'
data_mk1 = data[data['mata_kuliah'] == mk1]

# Filter data untuk mahasiswa yang mengambil mata kuliah "Struktur Data"
mk2 = 'Struktur Data'
data_mk2 = data[data['mata_kuliah'] == mk2]

# Gabungkan data menjadi satu DataFrame
data_box = pd.concat([data_mk1['nilai_akhir'], data_mk2['nilai_akhir']], axis=1)
data_box.columns = [mk1, mk2]

# Buat boxplot
plt.figure(figsize=(10, 6))
data_box.boxplot()
plt.title('Distribusi Nilai Akhir antara Mata Kuliah ' + mk1 + ' dan ' + mk2)
plt.ylabel('Nilai Akhir')
plt.xlabel('Mata Kuliah')
plt.grid(True)
plt.show()
```

# Scatter Plot

**Bagaimana hubungan antara kehadiran dan nilai akhir mahasiswa?**

<u>Interpretasi:</u>
- Amati distribusi titik.
- Jika ada korelasi positif, titik umumnya akan cenderung ke atas (kehadiran tinggi dikaitkan dengan nilai akhir tinggi).
- Jika ada korelasi negatif, titik akan cenderung ke bawah (kehadiran tinggi dikaitkan dengan nilai  akhir rendah).
- Distribusi acak tidak akan menunjukkan pola yang jelas.

```python
# Pilih variabel yang dibutuhkan
kehadiran = data["kehadiran"]
nilai_akhir = data["nilai_akhir"]

# Buat scatter plot
plt.figure(figsize=(8, 6))  # Atur ukuran plot (optional)
plt.scatter(kehadiran, nilai_akhir)

# Label sumbu
plt.xlabel("Kehadiran")
plt.ylabel("Nilai Akhir")

# Judul plot
plt.title("Hubungan antara Kehadiran dan Nilai Akhir")

# Tampilkan plot
plt.grid(True)  # Tambahkan grid (optional)
plt.show()
```

```python
# Membuat scatter plot nilai tugas dan nilai UTS
plt.figure(figsize=(8, 6))
plt.scatter(data['nilai_tugas'], data['nilai_uts'], color='blue', alpha=0.5)
plt.title('Scatter Plot Nilai Tugas vs Nilai UTS')
plt.xlabel('Nilai Tugas')
plt.ylabel('Nilai UTS')
plt.grid(True)
plt.show()
```

# Bar Chart

```python
# Menghitung jumlah mahasiswa untuk setiap mata kuliah
jumlah_mahasiswa_per_mata_kuliah = data['mata_kuliah'].value_counts()

# Membuat bar chart jumlah mahasiswa per mata kuliah
plt.figure(figsize=(10, 6))
jumlah_mahasiswa_per_mata_kuliah.plot(kind='bar')
plt.title('Jumlah Mahasiswa per Mata Kuliah')
plt.xlabel('Mata Kuliah')
plt.ylabel('Jumlah Mahasiswa')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Chart Jumlah Mahasiswa Berdasarkan Mata Kuliah dan Tahun**

```python
# Membuat pivot table untuk menghitung jumlah mahasiswa di setiap tahun untuk setiap mata kuliah
pivot_table = pd.pivot_table(data, index='mata_kuliah', columns='tahun_pengambilan', aggfunc='size', fill_value=0)

# Menampilkan pivot table
pivot_table
```

```python
# Menampilkan bar chart
pivot_table.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Jumlah Mahasiswa per Mata Kuliah dan Tahun')
plt.xlabel('Mata Kuliah')
plt.ylabel('Jumlah Mahasiswa')
plt.legend(title='Tahun')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# Pie Chart

Menggunakan pie chart untuk menampilkan proporsi jumlah mahasiswa yang mengambil setiap mata kuliah. 
Ini akan memberi kita gambaran visual tentang seberapa besar kontribusi setiap mata kuliah terhadap total jumlah mahasiswa.

```python
# Filter data untuk tahun terpilih
tahun = 2023
data_tahun = data[data['tahun_pengambilan'] == tahun]

# Menghitung jumlah mahasiswa untuk setiap mata kuliah di tahun terpilih
jumlah_mahasiswa_per_mata_kuliah = data_tahun['mata_kuliah'].value_counts()

# Membuat pie chart proporsi jumlah mahasiswa per mata kuliah di tahun 2024
plt.figure(figsize=(8, 8))
plt.pie(jumlah_mahasiswa_per_mata_kuliah, labels=jumlah_mahasiswa_per_mata_kuliah.index, autopct='%1.1f%%', startangle=140)
plt.title('Proporsi Jumlah Mahasiswa per Mata Kuliah di Tahun ' + str(tahun))
plt.axis('equal')  # Memastikan pie chart berbentuk lingkaran
plt.show()
```

Beberapa contoh kasus yang dapat diselesaikan menggunakan pie chart:
- **Proporsi Mahasiswa per Semester:** Pie chart dapat digunakan untuk menampilkan proporsi jumlah mahasiswa yang terdaftar di setiap semester. Ini akan memberikan gambaran tentang seberapa besar jumlah mahasiswa dalam tiap semester.
- **Proporsi Mahasiswa per Tahun:** Anda dapat menggunakan pie chart untuk menampilkan proporsi jumlah mahasiswa yang terdaftar di setiap tahun akademik. Ini akan membantu Anda melihat tren pertumbuhan jumlah mahasiswa dari tahun ke tahun.
- **Proporsi Mahasiswa per Indeks:** Pie chart dapat menampilkan proporsi jumlah mahasiswa yang mendapatkan indeks tertentu, seperti A, B, C, dst. Ini akan membantu Anda memahami distribusi kualitas kinerja akademik mahasiswa di berbagai indeks.
- **Proporsi Mahasiswa per Mata Kuliah:** Anda bisa membuat pie chart untuk menunjukkan proporsi jumlah mahasiswa yang mengambil setiap mata kuliah. Ini akan memberikan gambaran tentang popularitas setiap mata kuliah di antara mahasiswa.
- **Proporsi Mahasiswa per Tingkat Kehadiran:** Anda dapat menggunakan pie chart untuk menampilkan proporsi mahasiswa yang memiliki tingkat kehadiran tertentu, misalnya, hadir lebih dari 75%, hadir antara 50-75%, hadir antara 25-50%, dan hadir kurang dari 25%.
- **Proporsi Mahasiswa per Rentang Nilai Akhir:** Pie chart dapat menampilkan proporsi jumlah mahasiswa yang mendapatkan nilai akhir dalam rentang tertentu, misalnya, 0-50, 51-60, 61-70, 71-80, dan 81-100.

# Count Chart (Count Plot)

Baik count chart dan bar chart adalah jenis visualisasi data yang digunakan untuk menampilkan frekuensi atau jumlah data berdasarkan kategori. 
- Count Chart: Berfokus pada menampilkan jumlah data dalam setiap kategori secara absolut.
- Bar Chart: Berfokus pada membandingkan jumlah data antar kategori.

```python
nama_mk = "Pemrograman Berorientasi Objek"
data_pbo = data[data["mata_kuliah"] == nama_mk]

# Menghitung frekuensi indeks mata kuliah
frekuensi_indeks_matkul = data_pbo["indeks"].value_counts().sort_index()

# Membuat count chart
plt.figure(figsize=(10, 6))  # Atur ukuran plot (optional)
plt.bar(frekuensi_indeks_matkul.index, frekuensi_indeks_matkul.values)  # Buat batang
plt.xlabel("Indeks Mata Kuliah")  # Label sumbu X
plt.ylabel("Frekuensi")  # Label sumbu Y
plt.title("Distribusi Indeks Mata Kuliah " + nama_mk)  # Judul plot
plt.xticks(rotation=0, ha="right")  # Miringkan label sumbu X (optional)
plt.grid(axis="y")  # Tambahkan grid pada sumbu Y (optional)
plt.show()
```

```python
# Menyaring data untuk mata kuliah "Pemrograman Berorientasi Objek"
nama_mk = "Struktur Data"
data_pbo = data[data["mata_kuliah"] == nama_mk]

# Membuat count plot
sns.countplot(x="indeks", data=data_pbo, order=["A", "B", "C", "D", "E"])  # Urutkan indeks
plt.xlabel("Indeks Mata Kuliah " + nama_mk)
plt.ylabel("Frekuensi")
plt.title("Distribusi Indeks Mata Kuliah - " +nama_mk+ " (Urut A-E)")
plt.xticks(rotation=0, ha="right")  # Miringkan label sumbu X (optional)
plt.grid(axis="y")  # Tambahkan grid pada sumbu Y (optional)
plt.show()
```

# Joint Plot

Kasus: Hubungan antara Nilai Tugas dan Nilai UTS

Dalam kasus ini, kita ingin melihat apakah ada hubungan antara nilai rata-rata tugas mahasiswa dan nilai ujian tengah semester. Kita dapat menggunakan joint plot untuk menampilkan sebaran nilai-nilai keduanya dalam bentuk scatter plot, sambil memperhatikan garis regresi yang menunjukkan tren umum antara kedua variabel tersebut.

```python
# Filter data untuk memilih hanya nilai tugas dan nilai UTS
data_nilai = data[['nilai_tugas', 'nilai_uts']]

# Membuat joint plot
sns.jointplot(x='nilai_tugas', y='nilai_uts', data=data_nilai, kind='reg')
plt.xlabel('Nilai Tugas')
plt.ylabel('Nilai UTS')
plt.show()
```

```python
# Filter data untuk memilih hanya kehadiran dan nilai akhir
data_nilai_kehadiran = data[['kehadiran', 'nilai_akhir']]

# Membuat joint plot
sns.jointplot(x='kehadiran', y='nilai_akhir', data=data_nilai_kehadiran, kind='reg')
plt.xlabel('Jumlah Kehadiran')
plt.ylabel('Nilai Akhir')
plt.show()
```

# Contoh Grafik Lain

Line Plot

```python
# Mengelompokkan data berdasarkan semester dan menghitung rata-rata nilai
nilai_per_semester = data.groupby('semester')[['nilai_tugas', 'nilai_uts', 'nilai_uas', 'nilai_akhir']].mean()

# Membuat line plot perkembangan nilai
plt.figure(figsize=(10, 6))
plt.plot(nilai_per_semester.index, nilai_per_semester['nilai_tugas'], marker='o', label='Nilai Tugas')
plt.plot(nilai_per_semester.index, nilai_per_semester['nilai_uts'], marker='o', label='Nilai UTS')
plt.plot(nilai_per_semester.index, nilai_per_semester['nilai_uas'], marker='o', label='Nilai UAS')
plt.plot(nilai_per_semester.index, nilai_per_semester['nilai_akhir'], marker='o', label='Nilai Akhir')
plt.title('Perkembangan Nilai Per Semester')
plt.xlabel('Semester')
plt.ylabel('Nilai Rata-Rata')
plt.xticks(nilai_per_semester.index)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Perkembangan rata-rata nilai akhir untuk semua mata kuliah pada mahasiswa tertentu

# Filter data 
nim_mhs = 180001
data_mahasiswa = data[
    (data['nim'] == nim_mhs) & (data['semester'] >= 1)
]

# Hitung rata-rata nilai akhir per semester
rata_nilai_akhir_semester = data_mahasiswa.groupby('semester')['nilai_akhir'].mean().to_frame().reset_index()

# Buat line plot
sns.lineplot(x='semester', y='nilai_akhir', marker='o', data=rata_nilai_akhir_semester)
plt.title('Rata-rata Nilai Akhir Mahasiswa NIM ' + str(nim_mhs))
plt.xlabel('Semester')
plt.ylabel('Rata-rata Nilai Akhir')
plt.show()
```
