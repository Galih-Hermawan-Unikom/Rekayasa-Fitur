# Kasus 1. Analisis Sentimen pada Ulasan Restoran

Seorang pengusaha restoran ingin menganalisis sentimen pelanggan terhadap restorannya berdasarkan ulasan yang ditinggalkan di situs web. Dia memiliki dataset ulasan restoran yang berisi teks ulasan dan label sentimen (positif atau negatif) yang sesuai.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_selection import SelectKBest, chi2

# Muat dataset
data = {
    'Ulasan': [
        'Makanan di restoran ini sangat enak.',
        'Pelayanan di restoran ini sangat lambat.',
        'Restoran ini bersih dan nyaman.',
        'Harga makanan di restoran ini terlalu mahal.',
        'Makanan di restoran ini tidak enak.',
        'Pelayanan di restoran ini sangat baik.'
    ],
    'Sentimen': ['Positif', 'Negatif', 'Positif', 'Negatif', 'Negatif', 'Positif']
}
df = pd.DataFrame(data)

# Ekstraksi fitur
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['Ulasan'])

# Seleksi fitur dengan Chi-square Test
y = (df['Sentimen'] == 'Positif').astype(int)  # Konversi label ke numerik (1 untuk Positif, 0 untuk Negatif)
selector = SelectKBest(score_func=chi2, k=2)  # Pilih 2 fitur terbaik
X_selected = selector.fit_transform(X, y)

# Tampilkan fitur terpilih
selected_feature_names = [vectorizer.get_feature_names_out()[i] for i in selector.get_support(indices=True)]
print("Fitur Terpilih:", selected_feature_names)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_selection import chi2

# Data
data = {
    'Ulasan': [
        'Makanan di restoran ini sangat enak.',
        'Pelayanan di restoran ini sangat lambat.',
        'Restoran ini bersih dan nyaman.',
        'Harga makanan di restoran ini terlalu mahal.',
        'Makanan di restoran ini tidak enak.',
        'Pelayanan di restoran ini sangat baik.'
    ],
    'Sentimen': ['Positif', 'Negatif', 'Positif', 'Negatif', 'Negatif', 'Positif']
}

# 1. Membaca Data
df = pd.DataFrame(data)

# 2. Pra-Pemrosesan Data
# Tidak diperlukan dalam contoh ini karena menggunakan metode CountVectorizer().

# 3. Membuat Tabel Kontingensi
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['Ulasan'])
terms = vectorizer.get_feature_names_out()

# 4. Hitung Nilai Chi-Square
chi2_stat, p_values = chi2(X, df['Sentimen'])

# 5. Menampilkan Semua Hasil Perhitungan
result = pd.DataFrame({'Kata': terms, 'Chi-Square': chi2_stat, 'p-value': p_values})

print("Semua Hasil Perhitungan:")
print(result)

# 6. Seleksi Fitur dengan Filter
threshold = 0.5
selected_features = result[result['p-value'] < threshold]['Kata']

# 7. Menampilkan Hasil Filter
print("\nFitur Terpilih:")
print(selected_features.tolist())
```

```python
# 6. Seleksi Fitur dengan Filter
threshold = 0.8
selected_features = result[result['p-value'] < threshold]['Kata']

# 7. Menampilkan Hasil Filter
print("\nFitur Terpilih:")
print(selected_features.tolist())
```
