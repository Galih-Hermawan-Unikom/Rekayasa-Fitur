Kode `nltk.download('punkt')` dan `nltk.download('stopwords')` digunakan untuk mengunduh sumber daya data yang diperlukan untuk dua komponen penting dari library Natural Language Toolkit (NLTK):

**1. `nltk.download('punkt')`:**

- Perintah ini mengunduh data **tokenizer Punkt** untuk NLTK.
- Tokenizer Punkt adalah alat penting untuk melakukan **tokenisasi** teks, yang berarti memecah teks menjadi kata atau frasa individual.
- Tokenisasi adalah langkah fundamental dalam banyak tugas NLP, seperti analisis sentimen, terjemahan mesin, dan klasifikasi teks.
- Tokenizer Punkt secara khusus dilatih pada berbagai kumpulan teks (corpus) untuk mengidentifikasi batas kata secara akurat, terutama dalam bahasa seperti Bahasa Indonesia yang tidak memiliki pembatas kata yang jelas.

**2. `nltk.download('stopwords')`:**

- Perintah ini mengunduh **daftar stop word** untuk NLTK.
- Stop word adalah kata umum yang biasanya dianggap memiliki sedikit nilai semantik atau makna penting dalam bahasa tertentu.
- Contoh stop word termasuk "di", "dari", "pada", "yang", "untuk", dan lain-lain.
- Menghapus stop word dari teks bisa menjadi langkah pre-processing dalam tugas NLP untuk fokus pada konten yang lebih bermakna dan mengurangi noise (gangguan).
- NLTK menyediakan daftar stop word untuk banyak bahasa, termasuk Bahasa Indonesia.

# Menganalisis Kalimat

- Pre-processing
- Split kata dalam kalimat
- Menghitung frekuensi kata
- Menampilkan dalam grafik


```python
import re
import matplotlib.pyplot as plt

# Contoh kalimat
kalimat = "Saya senang belajar bahasa pemrograman Python. Python adalah bahasa yang mudah dipelajari dan digunakan. Ada banyak sumber daya yang tersedia untuk mempelajari Python, seperti buku, tutorial online, dan forum komunitas. Saya telah menemukan bahwa belajar Python sangat bermanfaat untuk pekerjaan dan proyek pribadi saya. Saya merekomendasikan Python kepada siapa pun yang ingin mempelajari bahasa pemrograman baru."

kalimat = """
Salah satu masalah paling signifikan pada Large Language Models adalah kecenderungan mereka untuk menghasilkan informasi yang salah atau mengada-ada, yang sering disebut sebagai "halusinasi". LLM bekerja dengan memprediksi urutan kata yang paling mungkin berdasarkan pola dalam data pelatihan mereka, bukan berdasarkan pemahaman faktual tentang dunia. Akibatnya, mereka bisa dengan yakin menyajikan klaim atau data yang sepenuhnya fiktif, membuat pengguna sulit membedakan kebenaran tanpa verifikasi eksternal. Masalah ini menjadi kritis terutama dalam aplikasi yang membutuhkan akurasi tinggi, seperti informasi medis, hukum, atau ilmiah, di mana kesalahan dapat memiliki konsekuensi serius.

Masalah penting lainnya adalah bias yang inheren dalam LLM. Karena model-model ini dilatih menggunakan data yang mencerminkan bias masyarakat yang ada (baik itu bias gender, ras, usia, atau lainnya), model tersebut dapat mempelajari dan mereplikasi bias ini dalam keluarannya. Hal ini bisa termanifestasi dalam stereotip, diskriminasi dalam rekomendasi, atau perlakuan yang tidak adil terhadap kelompok tertentu. Mengurangi bias dalam LLM adalah tantangan teknis dan etis yang kompleks, membutuhkan kurasi data yang cermat, pengembangan teknik debiasing, serta evaluasi berkelanjutan untuk memastikan keadilan dan kesetaraan dalam respons model.

Selain akurasi dan bias, ada juga tantangan besar terkait keamanan, transparansi, dan dampak sosial-ekonomi dari LLM. LLM dapat disalahgunakan untuk menghasilkan konten berbahaya seperti misinformasi, ujaran kebencian, spam, atau bahkan kode berbahaya. Mengontrol keluaran semacam ini sambil tetap memungkinkan fleksibilitas model adalah perjuangan yang berkelanjutan. Dari sisi transparansi, LLM sering dianggap sebagai "kotak hitam" karena sulit untuk memahami secara pasti mengapa model menghasilkan respons tertentu, yang menghambat debugging dan membangun kepercayaan. Terakhir, skala dan biaya operasional LLM sangat besar, membutuhkan sumber daya komputasi yang intensif, serta memunculkan kekhawatiran tentang dampak lingkungan dan potensi gangguan pasar kerja akibat otomatisasi yang didorong oleh teknologi ini.
"""
# Membersihkan teks
kalimat = kalimat.lower()
kalimat = re.sub(r'[^\w\s]', '', kalimat)

# Memotong kata dalam kalimat 
kata_kata = kalimat.split()

# Menghitung frekuensi kata
jumlah_kata = {}
for kata in kata_kata:
  if kata in jumlah_kata:
    jumlah_kata[kata] += 1
  else:
    jumlah_kata[kata] = 1

# Membuat daftar kata dan frekuensinya
label_kata = list(jumlah_kata.keys())
frekuensi_kata = list(jumlah_kata.values())

# Membuat grafik batang
plt.figure(figsize=(16, 6))
plt.bar(label_kata, frekuensi_kata, color='blue')
plt.xlabel('Kata')
plt.ylabel('Frekuensi')
plt.title('Frekuensi Kata dalam Kalimat')
plt.xticks(rotation=45)  # Miringkan label sumbu x untuk keterbacaan
plt.tight_layout()
plt.show()
```

Menyisipkan fungsi pendeteksian dan penghilangan stopwords menggunakan libray NLTK

```python
import re
import matplotlib.pyplot as plt
from nltk.corpus import stopwords

# Contoh kalimat
#kalimat = "Saya senang belajar bahasa pemrograman Python. Python adalah bahasa yang mudah dipelajari dan digunakan. Ada banyak sumber daya yang tersedia untuk mempelajari Python, seperti buku, tutorial online, dan forum komunitas. Saya telah menemukan bahwa belajar Python sangat bermanfaat untuk pekerjaan dan proyek pribadi saya. Saya merekomendasikan Python kepada siapa pun yang ingin mempelajari bahasa pemrograman baru."

# Membersihkan teks
kalimat = kalimat.lower()
kalimat = re.sub(r'[^\w\s]', '', kalimat)

# Memuat daftar stop word Bahasa Indonesia
stop_words = set(stopwords.words('indonesian'))

# Pisahkan teks menjadi kata-kata (tokenisasi)
kata_kata = kalimat.split()

# Filter stop word secara manual (tanpa fungsi)
kata_tanpa_stopword = [kata for kata in kata_kata if kata not in stop_words]

# Gabungkan kembali kata menjadi kalimat
teks_final = ' '.join(kata_tanpa_stopword)

# Menghitung frekuensi kata
jumlah_kata = {}
for kata in kata_tanpa_stopword:
  if kata in jumlah_kata:
    jumlah_kata[kata] += 1
  else:
    jumlah_kata[kata] = 1

# Membuat daftar kata dan frekuensinya
label_kata = list(jumlah_kata.keys())
frekuensi_kata = list(jumlah_kata.values())

# Membuat grafik batang
plt.figure(figsize=(12, 6))
plt.bar(label_kata, frekuensi_kata, color='blue')
plt.xlabel('Kata')
plt.ylabel('Frekuensi')
plt.title('Frekuensi Kata dalam Kalimat Tanpa Stopword')
plt.xticks(rotation=45)  # Miringkan label sumbu x untuk keterbacaan
plt.tight_layout()
plt.show()
```

Membuat **wordcloud** menggunakan library Wordcloud.

- Wordcloud (atau bisa disebut cloud kata) adalah sebuah representasi visual dari kumpulan kata-kata. 
- Bayangkan seperti awan yang bentuknya dibuat oleh kumpulan kata. 
- Semakin sering sebuah kata muncul dalam kumpulan tersebut, maka ukuran kata itu akan semakin besar dan semakin terlihat menonjol di dalam wordcloud.
- Ini seperti cara sederhana untuk melihat sekilas "inti" atau topik utama dari suatu kumpulan kata.

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Membuat awan kata
wordcloud = WordCloud(width=800, height=400).generate(teks_final)

# Menampilkan awan kata
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
