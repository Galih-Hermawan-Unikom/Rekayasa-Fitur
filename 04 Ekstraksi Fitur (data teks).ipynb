**Contoh ekstraksi fitur pada teks dari SMS**

Dapat diterapkan pada kasus: <u>klasifikasi SMS sebagai spam atau bukan spam berdasarkan teks SMS</u>

**Langkah-langkah Ekstraksi Fitur dari Data Teks**

1. **Preprocessing Teks**:
    - Hapus tanda baca, angka, dan simbol.
    - Konversi teks menjadi huruf kecil.
    - Hapus spasi berlebihan.
    - Ubah kata yang disingkat menjadi kata lengkap (misalnya, "bgt" menjadi "banget").

2. **Ekstraksi Fitur**:
    - **Panjang SMS**: Hitung jumlah karakter atau kata dalam SMS.
    - **Frekuensi Kata Kunci**: Hitung frekuensi kemunculan kata-kata kunci yang terkait dengan spam, seperti "promo", "gratis", "diskon", "menang", dll.
    - **Kata-kata Kapital**: Hitung jumlah kata yang ditulis dengan huruf kapital.
    - **Tanda Seru**: Hitung jumlah tanda seru (!).
    - **Tautan**: Hitung jumlah tautan dalam SMS.

3. **Representasi Fitur**:
    - Ubah fitur yang diekstrak menjadi representasi numerik.
    - Contohnya, gunakan vektor dengan elemen yang mewakili panjang SMS, frekuensi kata kunci, jumlah kata kapital, jumlah tanda seru, dan jumlah tautan.



```python
import re

# Contoh teks SMS
sms = [
    "PROMO! Dapatkan diskon 50% untuk semua produk di toko kami. Beli sekarang dan hemat banyak! http://tes.com",
    "Halo, ini dari bank Anda. Akun Anda telah diblokir karena aktivitas mencurigakan. Silakan hubungi customer service kami segera.",
    "Selamat! Anda telah memenangkan hadiah utama dalam undian kami. Hubungi [nomor telepon] untuk klaim hadiah Anda."
]

# Fungsi untuk mengubah singkatan menjadi kata lengkap
def ubah_singkatan(teks):
    # Daftar singkatan yang ingin diubah
    daftar_singkatan = {
        "sdg": "sedang",
        "km": "kamu",
        "sy": "saya",
        "lg": "lagi",
        "gak": "tidak",
        "bgt": "banget",
        # Tambahkan singkatan lain jika diperlukan
    }
    
    # Ubah singkatan menjadi kata lengkap
    for singkatan, kata_lengkap in daftar_singkatan.items():
        teks = teks.replace(singkatan, kata_lengkap)
    
    return teks

# Preprocessing teks
def praproses_teks(teks):
    # Ubah kata yang disingkat menjadi kata lengkap (implementasi fungsi untuk ini)
    teks = ubah_singkatan(teks)
    return teks

teks_praproses_sms = [praproses_teks(teks_sms) for teks_sms in sms]

# Ekstraksi fitur
def ekstraksi_fitur(teks):
    # Panjang SMS
    panjang = len(teks)
    
    # Frekuensi kata kunci
    hitung_kata_kunci = {}
    for kata_kunci in ["promo", "gratis", "diskon", "menang"]:
        hitung_kata_kunci[kata_kunci] = teks.count(kata_kunci)
    
    # Kata-kata kapital
    jumlah_kata_kapital = sum([1 for kata in teks.split() if kata.isupper()])
    
    # Tanda seru
    jumlah_tanda_seru = teks.count("!")
    
    # Tautan
    jumlah_tautan = len(re.findall(r'(https?://\S+)', teks))
    
    # Representasi fitur
    fitur = [panjang, hitung_kata_kunci["promo"], hitung_kata_kunci["gratis"], hitung_kata_kunci["diskon"], hitung_kata_kunci["menang"], jumlah_kata_kapital, jumlah_tanda_seru, jumlah_tautan] # Menambahkan jumlah tautan sebagai fitur
    return fitur

# Ekstraksi fitur dan praproses teks
fitur = []
jumlah_kata_kapital = []
jumlah_tanda_seru = []

for teks_sms in sms:
    # Praproses teks
    teks_praproses_sms = praproses_teks(teks_sms)
    
    # Ekstraksi fitur
    fitur_sms = ekstraksi_fitur(teks_praproses_sms)
    
    # Jumlah kata kapital
    jumlah_kata_kapital_sms = sum([1 for kata in teks_praproses_sms.split() if kata.isupper()])
    jumlah_kata_kapital.append(jumlah_kata_kapital_sms)
    
    # Jumlah tanda seru
    jumlah_tanda_seru_sms = teks_praproses_sms.count("!")
    jumlah_tanda_seru.append(jumlah_tanda_seru_sms)
    
    # Tambahkan fitur ke daftar fitur
    fitur.append(fitur_sms)

# Cetak hasil ekstraksi fitur
print("Hasil ekstraksi fitur:")
"""
1. Panjang SMS
2. Frekuensi kata kunci "promo"
3. Frekuensi kata kunci "gratis"
4. Frekuensi kata kunci "diskon"
5. Frekuensi kata kunci "menang"
6. Jumlah kata kapital
7. Jumlah tanda seru
8. Jumlah tautan
"""
for i, fitur_sms in enumerate(fitur):
    print(f"SMS {i+1}: {fitur_sms}")
```

---

Teknik ekstraksi fitur pada teks secara manual dapat diterapkan pada berbagai jenis teks, tidak hanya SMS. Berikut beberapa contoh:

**1. Analisis sentimen:**

* **Teks:** Ulasan produk, komentar media sosial, tweet.
* **Fitur:**
    * Kata-kata positif: "bagus", "senang", "puas".
    * Kata-kata negatif: "buruk", "kecewa", "jelek".
    * Penggunaan tanda seru (!).
    * Kata-kata kapital.
    * Emoticon (senyum, sedih).

**2. Klasifikasi topik:**

* **Teks:** Berita online, artikel blog, email.
* **Fitur:**
    * Kata kunci yang terkait dengan topik tertentu (misalnya, "olahraga", "kesehatan", "politik").
    * Kata benda dan frasa nomina yang mendeskripsikan topik.
    * Kata kerja yang menunjukkan aksi terkait topik.

**3. Klasifikasi dokumen:**

* **Teks:** Email spam, resume pelamar, dokumen hukum.
* **Fitur:**
    * Kata-kata yang umum ditemukan pada jenis dokumen tertentu (misalnya, "diskon", "gratis" untuk spam, "keahlian", "pengalaman" untuk resume).
    * Panjang dokumen.
    * Format penulisan (formal, informal).
    * Penggunaan struktur tertentu (judul, subjudul, tanggal).

**4. Ringkasan teks otomatis:**

* **Teks:** Artikel berita, laporan keuangan, transkrip wawancara.
* **Fitur:**
    * Kalimat yang mengandung kata kunci atau entitas penting.
    * Kalimat dengan struktur subjek-predikat yang lengkap.
    * Skor TF-IDF (Term Frequency-Inverse Document Frequency) untuk mengukur pentingnya kata dalam dokumen.

**5. Terjemahan mesin:**

* **Teks:** Kalimat atau paragraf dalam satu bahasa.
* **Fitur:**
    * Part-of-speech (POS) tagging untuk mengidentifikasi jenis kata (kata benda, kata kerja, dll.).
    * Struktur kalimat (subjek, objek, verba).
    * Frasa dan idiom yang umum digunakan.

**Hal penting yang perlu diingat dalam ekstraksi fitur secara manual:**

* **Relevansi:** Pastikan fitur yang diekstrak relevan dengan tugas yang ingin Anda lakukan.
* **Keakuratan:** Definisikan dan ekstrak fitur dengan cara yang konsisten dan akurat.
* **Keterbatasan:** Ekstraksi manual bisa memakan waktu dan tidak berskala dengan baik untuk data yang besar.

**Selain teknik di atas, ada juga teknik ekstraksi fitur otomatis yang menggunakan machine learning. Namun, teknik manual tetap berguna untuk memahami struktur dan isi teks, serta membantu dalam pemilihan fitur yang tepat untuk model machine learning.**
