Pada gambar hitam putih berukuran 10x10 piksel, deteksi tepi mungkin tidak diperlukan karena beberapa alasan:

1. Resolusi Rendah:

- Ukuran gambar yang kecil (10x10 piksel) berarti gambar hanya memiliki 100 piksel.
- Dengan resolusi rendah ini, detail gambar mungkin tidak cukup untuk diidentifikasi dengan baik.
- Deteksi tepi pada gambar dengan detail yang minim dapat menghasilkan hasil yang tidak akurat atau tidak bermakna.

2. Kontras Tinggi:

- Gambar hitam putih biasanya memiliki kontras tinggi antara area terang dan gelap.
- Dalam banyak kasus, tepi gambar sudah terlihat jelas tanpa memerlukan deteksi tepi tambahan.
- Menerapkan algoritma deteksi tepi pada gambar kontras tinggi dapat memperkuat tepi yang sudah ada secara berlebihan dan menghasilkan artefak.

3. Kehilangan Informasi:

- Deteksi tepi hanya fokus pada lokasi perubahan intensitas, mengabaikan informasi intensitas piksel itu sendiri.
- Pada gambar kecil dengan detail terbatas, kehilangan informasi intensitas ini dapat signifikan.

**Matriks Korelasi dan Koefisien Korelasi untuk Membandingkan Dua Gambar**

**Koefisien korelasi** adalah nilai numerik antara -1 dan 1 yang menunjukkan kekuatan dan arah hubungan antara dua variabel.

Saat membandingkan dua gambar, matriks korelasi dan koefisien korelasi dapat digunakan untuk mengukur kesamaan atau perbedaan antara gambar tersebut.

**Matriks korelasi** akan menunjukkan koefisien korelasi antara setiap pasangan piksel di kedua gambar. Koefisien korelasi yang tinggi untuk sepasang piksel menunjukkan bahwa piksel tersebut memiliki nilai intensitas yang serupa di kedua gambar.

**Koefisien korelasi** keseluruhan antara kedua gambar dapat dihitung dengan rata-rata semua koefisien korelasi individu dalam matriks. Koefisien korelasi keseluruhan yang tinggi menunjukkan bahwa kedua gambar secara keseluruhan memiliki kesamaan intensitas piksel yang tinggi.

**Interpretasi:**

- Koefisien korelasi positif: Menunjukkan bahwa gambar umumnya memiliki intensitas piksel yang serupa, dengan area terang dan gelap yang sesuai di kedua gambar.
- Koefisien korelasi negatif: Menunjukkan bahwa gambar umumnya memiliki intensitas piksel yang berlawanan, dengan area terang di satu gambar yang sesuai dengan area gelap di gambar lain.
- Koefisien korelasi mendekati 0: Menunjukkan bahwa tidak ada hubungan linier yang jelas antara intensitas piksel di kedua gambar.

- Berikut contoh fungsi untuk memeriksa kemiripan dua gambar berdasarkan matriks korelasi.
- Matriks korelasi diukur berdasarkan nilai piksel gambar hasil normalisasi tanpa melalui thresholding.
- Bisakah Anda bereksperimen, apakah ada perbedaan jika menggunakan gambar hasil **thresholding**?

```python
def periksa_kemiripan_gambar(gambar1, gambar2):
    """
    Fungsi ini memeriksa kemiripan antara dua gambar berdasarkan matriks korelasi dan memberikan status kemiripan.

    Args:
    gambar1: Matriks yang mewakili intensitas piksel gambar pertama.
    gambar2: Matriks yang mewakili intensitas piksel gambar kedua.

    Returns:
    skor_kemiripan: Nilai numerik antara -1 dan 1 yang menunjukkan tingkat kemiripan antara kedua gambar.
    status_kemiripan: String yang menunjukkan kategori status kemiripan berdasarkan skor_kemiripan.
    """

    img1 = cv2.imread(gambar1, cv2.IMREAD_GRAYSCALE)
    img2 = cv2.imread(gambar2, cv2.IMREAD_GRAYSCALE)

    # Normalisasi gambar
    img1_norm = (img1 - img1.min()) / (img1.max() - img1.min())
    img2_norm = (img2 - img2.min()) / (img2.max() - img2.min())

    # Hitung matriks korelasi
    matriks_korelasi = np.corrcoef(img1_norm.flatten(), img2_norm.flatten())

    # Hitung nilai rata-rata
    skor_korelasi_rata_rata = np.mean(matriks_korelasi)

    # Tetapkan status kemiripan
    if skor_korelasi_rata_rata >= 0.9:
        status_kemiripan = "Mirip Sempurna"
    elif 0.7 <= skor_korelasi_rata_rata < 0.9:
        status_kemiripan = "Agak Mirip"
    elif 0.5 <= skor_korelasi_rata_rata < 0.7:
        status_kemiripan = "Kurang Mirip"
    else:
        status_kemiripan = "Tidak Mirip"

    # Kembalikan skor kemiripan dan status
    return round(skor_korelasi_rata_rata,2), status_kemiripan

folder = "angka"
fail_gambar1 = "2_2.jpg"
fail_gambar2 = "4_3.jpg"

skor, status = periksa_kemiripan_gambar(folder + "/" + fail_gambar1, folder + "/" + fail_gambar2)
print(f"Skor kemiripan: {skor}")
print(f"Status kemiripan: {status}")
```

Membuat Matriks Korelasi untuk Semua Gambar disertai Heatmap

```python
import numpy as np  # Pustaka untuk komputasi numerik
import pandas as pd  # Pustaka untuk analisis data
from PIL import Image  # Pustaka untuk memproses gambar
import os  # Pustaka untuk mengakses sistem file
import matplotlib.pyplot as plt  # Pustaka untuk visualisasi data
import seaborn as sns  # Pustaka untuk visualisasi statistik

# Fungsi untuk menghitung matriks korelasi
def hitung_matriks_korelasi(folder_gambar):
  """
  Fungsi ini menghitung matriks korelasi untuk gambar dalam folder yang diberikan.

  Args:
    folder_gambar: Jalur ke folder yang berisi gambar.

  Returns:
    Matriks korelasi dan label gambar.
  """

  # Dapatkan daftar nama file gambar
  daftar_file_gambar = [os.path.join(folder_gambar, f) for f in os.listdir(folder_gambar) if os.path.isfile(os.path.join(folder_gambar, f))]

  # Periksa apakah ada gambar
  if not daftar_file_gambar:
    raise Exception("Tidak ada gambar yang ditemukan di folder.")

  # Baca dan representasikan setiap gambar sebagai vektor numerik
  gambar_vektor = []
  for file_gambar in daftar_file_gambar:
    gambar = Image.open(file_gambar).convert('L')  # Konversi ke grayscale
    data_gambar = np.array(gambar).flatten()
    gambar_vektor.append(data_gambar)

  # Hitung matriks korelasi
  matriks_korelasi = np.zeros((len(gambar_vektor), len(gambar_vektor)))
  for i in range(len(gambar_vektor)):
    for j in range(len(gambar_vektor)):
      if i == j:
        matriks_korelasi[i, j] = 1.0
      else:
        matriks_korelasi[i, j] = np.corrcoef(gambar_vektor[i], gambar_vektor[j])[0][1]

  # Ubah label dari nama file gambar
  label_gambar = [os.path.basename(os.path.splitext(f)[0]) for f in daftar_file_gambar]

  return matriks_korelasi, label_gambar

# Jalankan fungsi dan dapatkan matriks korelasi dan label
matriks_korelasi, label_gambar = hitung_matriks_korelasi("angka")

# Buat heatmap dengan anotasi
plt.figure(figsize=(10, 6))  # Atur ukuran gambar heatmap
sns.heatmap(matriks_korelasi, cmap='coolwarm', annot=True, fmt=".2f", xticklabels=label_gambar, yticklabels=label_gambar)
plt.title("Heatmap Korelasi Semua Gambar")  # Atur judul heatmap
plt.xlabel('Nama Gambar')  # Atur label sumbu x
plt.ylabel('Nama Gambar')  # Atur label sumbu y
plt.show()
```
