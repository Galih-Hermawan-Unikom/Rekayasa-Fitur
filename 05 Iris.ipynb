**Dataset Iris** adalah kumpulan data multivariat yang terkenal dalam ilmu komputer dan statistik. Dataset ini pertama kali diperkenalkan oleh Ronald Fisher dalam makalahnya tahun 1936, "The Use of Multiple Measurements in Taxonomic Problems". Dataset ini digunakan untuk mengilustrasikan konsep analisis diskriminan linier (LDA) dan telah menjadi tolok ukur untuk berbagai algoritma klasifikasi mesin.

**Karakteristik Dataset:**

* **Jumlah data:** 150 sampel
* **Kelas:** 3 kelas (Iris setosa, Iris versicolor, dan Iris virginica)
* **Fitur:** 4 fitur numerik (sepal length, sepal width, petal length, petal width)

**Sumber Data:**

* Dataset Iris tersedia secara bebas dari berbagai sumber online, termasuk:
    * UCI Machine Learning Repository: [https://archive.ics.uci.edu/dataset/53/iris](https://archive.ics.uci.edu/dataset/53/iris)
    * Kaggle: [https://www.kaggle.com/datasets/uciml/iris](https://www.kaggle.com/datasets/uciml/iris)
* Dataset ini juga dapat dimuat secara langsung menggunakan pustaka Python seperti `scikit-learn` dan `pandas`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

# Memuat dataset Iris
data_iris = load_iris()

# Konversi dataset menjadi DataFrame
iris_df = pd.DataFrame(data=data_iris.data, columns=data_iris.feature_names)

# Menambahkan kolom spesies ke dalam DataFrame
iris_df['spesies'] = data_iris.target

iris_df
```

Visualisasi **histogram** fitur (panjang sepal, lebar sepal, panjang petal, lebar petal) untuk memahami distribusi data.

```python
# Visualisasi histogram untuk fitur petal length
plt.figure(figsize=(8, 6))
sns.histplot(data=iris_df, x='petal length (cm)', hue='spesies', kde=True, palette='Set1')
plt.title('Histogram Panjang Kelopak')
plt.xlabel('Panjang Kelopak (cm)')
plt.ylabel('Frekuensi')
plt.grid(True)
plt.show()

# Visualisasi histogram untuk fitur petal width
plt.figure(figsize=(8, 6))
sns.histplot(data=iris_df, x='petal width (cm)', hue='spesies', kde=True, palette='Set1')
plt.title('Histogram Lebar Kelopak')
plt.xlabel('Lebar Kelopak (cm)')
plt.ylabel('Frekuensi')
plt.grid(True)
plt.show()
```

**Scatter plot** berikut menunjukkan hubungan antara panjang dan lebar kelopak bunga Iris, diwarnai berdasarkan spesiesnya. Hal ini membantu memvisualisasikan bagaimana panjang dan lebar kelopak dapat membantu membedakan antara spesies Iris yang berbeda.

```python
# Visualisasi scatter plot untuk panjang dan lebar kelopak
plt.figure(figsize=(8, 6))
sns.scatterplot(x='petal length (cm)', y='petal width (cm)', data=iris_df, hue='spesies', palette='Set1')
plt.title('Scatter Plot Panjang dan Lebar Kelopak')
plt.xlabel('Panjang Kelopak (cm)')
plt.ylabel('Lebar Kelopak (cm)')
plt.grid(True)
plt.show()
```

```python
# Menambahkan kolom spesies ke dalam DataFrame dengan nama spesies
iris_df['spesies'] = data_iris.target
iris_df['spesies'] = iris_df['spesies'].map({0: 'Iris Setosa', 1: 'Iris Versicolor', 2: 'Iris Virginica'})

# Visualisasi scatter plot untuk panjang dan lebar kelopak
plt.figure(figsize=(8, 6))
sns.scatterplot(x='petal length (cm)', y='petal width (cm)', data=iris_df, hue='spesies', palette='Set1')
plt.title('Scatter Plot Panjang dan Lebar Kelopak')
plt.xlabel('Panjang Kelopak (cm)')
plt.ylabel('Lebar Kelopak (cm)')
plt.grid(True)
plt.show()
```

Berikut adalah **heatmap** yang menunjukkan korelasi antara semua fitur numerik dalam dataset Iris. Warna-warna dalam heatmap membantu memvisualisasikan kekuatan dan arah korelasi antar fitur. Hal ini dapat membantu dalam memahami hubungan antar fitur dan mengidentifikasi fitur yang mungkin saling terkait.

```python
# Visualisasi korelasi antar fitur menggunakan heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(iris_df.corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap Korelasi Fitur dalam Dataset Iris')
plt.show()
```

**Pair plot** membantu memvisualisasikan hubungan antar semua pasangan fitur dalam dataset Iris secara simultan. Hal ini dapat membantu dalam mengidentifikasi pola dan tren dalam hubungan antar fitur, serta dalam memahami bagaimana fitur-fitur tersebut mungkin saling terkait.

```python
# Visualisasi hubungan antar pasangan fitur menggunakan pair plot
sns.pairplot(iris_df, hue='spesies', palette='Set1')
plt.title('Plot Pasangan dari Dataset Iris')
plt.show()
```

Penggunaan **box plot** untuk melihat distribusi dan variabilitas data serta identifikasi pencilan (outliers).

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

# Memuat dataset Iris
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Membuat box plot untuk setiap fitur
plt.figure(figsize=(10, 6))
sns.boxplot(data=iris_df)
plt.title('Box Plot untuk Distribusi dan Variabilitas Data Iris')
plt.xlabel('Fitur')
plt.ylabel('Nilai')
plt.xticks(rotation=45)  # Memutar label sumbu x agar lebih mudah dibaca
plt.show()
```
